# Data inspection
---

```python
import json
import os
from pathlib import Path
from datetime import datetime, timedelta
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from rctoolbox.utils.project_root import find_project_root

project_root = find_project_root()
os.chdir(project_root)
# print(f"Current working directory: {os.getcwd()}")

DATA_DIR  = project_root / "examples" / "data"
```

## AIBF
---

```python
hems_id = "AIBF"

data_path = DATA_DIR / f'{hems_id}_data.pkl'
DATA = pd.read_pickle(data_path)
DATA.keys()
```

```python
SEASON = 'w2425'
season = DATA[SEASON]
df = season['df']

print('season:        ', SEASON)
print('df shape:      ', df.shape)
print('time range:    ', df.index.min(), '->', df.index.max())
print(df.columns, df.index.min(), df.index.max(), df.index.freq)

# PLOT
# t1 = datetime(2023,12,1)
# t2 = t1 + timedelta(days=5*30)

t1 = df.index.min()
t2 = df.index.max()

df_ = df.loc[t1:t2]
df_.plot(subplots=True, figsize=(10, 6), title=f"{SEASON} data from {t1.date()} to {t2.date()}")
plt.tight_layout()
```
