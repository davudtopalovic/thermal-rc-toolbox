# 3R3C: air + internal mass + evaporator coil, COP(T_out) and outdoor link — heat balances and state-space matrices

```
                     Phi = Phi_HC (1 + beta (T_out - 10))
                                    |
   T_a ---R_ia--- T_i ---g(s)------ T_e
   T_out-R_iout--/ |  \              |
                  C_i  R_im          C_e
                         \
                          T_m --- C_m
```

| symbol | unit | meaning |
|---|---|---|
| $T_i$ | °C | freezer air (state, measured, thermostat sensor) |
| $T_m$ | °C | internal mass: stored products, shelves, inner surfaces (state, hidden) |
| $T_e$ | °C | evaporator coil incl. fins, frost and refrigerant (state, hidden) |
| $T_a$ | °C | neighbouring cold room, `Cold room temperature` (input) |
| $T_{out}$ | °C | outdoor air, `temperature_2m` (input) |
| $\Phi_{HC}$ | W | cooling power, $\Phi_{HC} = -P_{el}$ (input; effective values absorb the COP) |
| $s$ | – | compressor/fan on-fraction of the interval (input, `on`) |

Free parameters: `C_i`, `R_ia`, `R_im`, `C_m`, `R_ie_on`, `R_ie_off`, `C_e`, `R_iout`, `beta`. Registry key `fz_3R3C_im_ev_cop_temperature`, class
`Freezer3R3C_IM_EV_COP` (a fixed configuration of `FreezerNetwork`, see `freezer_models/network.py`).
The air–coil coupling is fan-gated: $g(s) = s/R_{ie}^{on} + (1-s)/R_{ie}^{off}$ with $s$ the compressor on-fraction, so the matrices depend on $s$ (switched linear model, affine in $s$).
All cooling acts on the coil node: the air is cooled only through the coil (forced-air evaporator).



````python
import numpy as np
import sympy as sp
from IPython.display import display, Math

sp.init_printing()

def show(lhs, expr):
    display(Math(lhs + " = " + sp.latex(expr)))
````

## Heat balances

$$C_{i} \frac{dT_i}{dt} = \frac{T_{a} - T_{i}}{R_{ia}} + \frac{- T_{i} + T_{m}}{R_{im}} + \left(T_{e} - T_{i}\right) \left(\frac{s}{R_{ie}^{on}} + \frac{1 - s}{R_{ie}^{off}}\right) + \frac{- T_{i} + T_{out}}{R_{i,out}}$$

$$C_{m} \frac{dT_m}{dt} = \frac{T_{i} - T_{m}}{R_{im}}$$

$$C_{e} \frac{dT_e}{dt} = \left(- T_{e} + T_{i}\right) \left(\frac{s}{R_{ie}^{on}} + \frac{1 - s}{R_{ie}^{off}}\right) + \Phi_{HC} \left(\beta \left(T_{out} - 10\right) + 1\right)$$

````python
T_i, T_m, T_e, T_a, T_out, T_ev, Phi_HC = sp.symbols("T_i T_m T_e T_a T_out T_ev Phi_HC", real=True)
R_ia, R_im, R_ie_on, R_ie_off, R_ea, R_iout = sp.symbols("R_ia R_im R_ie_on R_ie_off R_ea R_iout", positive=True)
C_i, C_m, C_e = sp.symbols("C_i C_m C_e", positive=True)
beta = sp.Symbol("beta", real=True)
s = sp.Symbol("s", nonnegative=True)   # compressor on-fraction
dT_i, dT_m, dT_e = sp.symbols(r"\dot{T_i} \dot{T_m} \dot{T_e}", seq=True)

balances = [
    sp.Eq(C_i * dT_i, (T_a - T_i) / R_ia + (T_m - T_i) / R_im + (s / R_ie_on + (1 - s) / R_ie_off) * (T_e - T_i) + (T_out - T_i) / R_iout),
    sp.Eq(C_m * dT_m, (T_i - T_m) / R_im),
    sp.Eq(C_e * dT_e, (s / R_ie_on + (1 - s) / R_ie_off) * (T_i - T_e) + Phi_HC * (1 + beta * (T_out - 10))),
]
for b in balances:
    display(b)

sol = sp.solve(balances, [dT_i, dT_m, dT_e], dict=True)[0]
f = sp.Matrix([sp.expand(sol[dT_i]), sp.expand(sol[dT_m]), sp.expand(sol[dT_e])])
for name, fx in zip(['T_i', 'T_m', 'T_e'], f):
    show(r"\dot{" + name + "}", fx)
````

## State-space form

The balances are linear in the states and in the input vector

$$u = [\Phi_{HC}, T_{a}, \Phi_{HC} \left(T_{out} - 10\right), T_{out}]$$

(the order of `INPUT_NAMES` in the class; the COP term is linear through the extra input $\Phi_{HC}(T_{out}-10)$), so $\dot x = A(s)\,x + B(s)\,u$ with $A = \partial f/\partial x$ and $B = \partial f/\partial u$.

````python
x = sp.Matrix([T_i, T_m, T_e])
w = sp.Symbol("w")   # stands for the product input Phi_HC*(T_out - 10)
u_expr = [Phi_HC, T_a, Phi_HC * (T_out - 10), T_out]
u_sym = [Phi_HC, T_a, w, T_out]
f_w = f.subs(Phi_HC * beta * T_out, beta * (w + 10 * Phi_HC)).expand() if True else f
f_w = sp.expand(f_w)
A = f_w.jacobian(x)
B = f_w.jacobian(sp.Matrix(u_sym))
show("A", sp.simplify(A))
show("B", sp.simplify(B))
residual = sp.simplify((A * x + B * sp.Matrix(u_sym) - f_w))
assert residual == sp.zeros(*f.shape), residual
print("A x + B u == f  ✓  (u =", u_sym, ")")
````

## Check against the toolbox implementation

The symbolic matrices are evaluated for random parameter values (and on/off states) and compared with
`calc_matrices(s)` of the class. The cells after that check the discretization and the limits numerically:

- exact ZOH (`zoh_discretize`, used for identification) against a forward-Euler simulation with a 1000× finer step;
- the steady state of a long simulation with constant inputs against $x_{ss} = -A^{-1}B u$;
- reduction to the parent model when the extra element is removed.

````python
from rctoolbox.models.rc.freezer_models.fz_3r3c_im_ev_cop.model import Freezer3R3C_IM_EV_COP
from rctoolbox.models.rc.freezer_models._checks import random_params

rng = np.random.default_rng(0)
symmap = {"R_ia": R_ia, "R_im": R_im, "R_ie_on": R_ie_on, "R_ie_off": R_ie_off, "R_ea": R_ea, "R_iout": R_iout,
          "C_i": C_i, "C_m": C_m, "C_e": C_e, "beta": beta}
model = Freezer3R3C_IM_EV_COP()
print("states", model.STATE_NAMES, "| inputs", model.INPUT_NAMES, "| switched", model.SWITCHED)
for _ in range(5):
    p = random_params(model, rng)
    model.set_params(p)
    for sv in (0.0, 0.5, 1.0):
        A_num, B_num = model.calc_matrices(sv)
        subs = {symmap[k]: v for k, v in p.items()} | {s: sv}
        assert np.allclose(np.array(A.subs(subs), float), A_num, rtol=1e-10, atol=1e-15)
        assert np.allclose(np.array(B.subs(subs), float), B_num, rtol=1e-10, atol=1e-15)
print("symbolic A, B == Freezer3R3C_IM_EV_COP.calc_matrices(s)  ✓")
````
